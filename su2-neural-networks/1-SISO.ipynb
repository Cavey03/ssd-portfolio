# SISO - A league of legends win probability estimation based on creep score.


Creep score refers to the minions you kill for gold, and higher= better
In competitive League of Legends, Creep Score (CS) at the 10-minute mark reflects how efficiently a player has farmed minions. It is a strong indicator of economic and mechanical advantage

This model is a single-input, single-output neural network that takes one stat, CS@10, and outputs a single estimated win probability




creep.jpg

```python
weight = 0.01

def neural_network(input, weight):
    prediction = input * weight
    return prediction

# CS at the 10-minute mark for 4 sample games.

cs_at_10 = [75, 68, 82, 60]

input = cs_at_10[0]
pred = neural_network(input, weight)
print(pred)

```

# **Understanding Output**

First game the CS@10 gives a prediction of 0.75, with an input of 75 creeps at the 10 minute mark.

Higher CS means higher predictions.
Note! - errored model, single multiplication with no bias, means very high input could push the output >1.0 which doesnt make sense

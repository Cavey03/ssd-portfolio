# SIMO using Gold diff@10 to predict {win probability, expected kill lead, objective control chance}.

 Gold difference at 10 minutes is one of the strongest single predictors of a team's overall trajectory in a game.

 This notebook takes one input, gold diff@10 and predicts win probability, expected kill lead, and objective control chance

```python
def ele_mul(number, vector):
    output = [0, 0, 0]
    assert(len(output) == len(vector))
    for i in range(len(vector)):
        output[i] = number * vector[i]
    return output

weights = [0.0005, 0.004, 0.0003]  # [win_prob, kill_lead, objective_control]

def neural_network(input, weights):
    pred = ele_mul(input, weights)
    return pred

gold_diff_10 = [450, -200, 900, -600]

input = gold_diff_10[0]
pred = neural_network(input, weights)
print(pred)
```

**Understanding the Output**

In game 1 (+450 gold lead) the model predicts win probability ≈ 0.23, expected kill lead ≈ 1.8 kills, objective control chance ≈ 0.14.

A gold lead of 450 is a decent advantage and the win prediction being only 0.23 could be accurate.

When we look at game 2 (-200 gold deficit) the same weights give [-0.1, -0.8, -0.06], that is a negative win prediction which is meaningless. Cause because there is no bias term and no activation function to make sure outputs are in a valid range
